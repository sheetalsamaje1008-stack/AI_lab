```python
from itertools import product

def entails(kb, query, symbols):
    models = []

    for values in product([False, True], repeat=len(symbols)):
        model = dict(zip(symbols, values))

        if all(sentence(model) for sentence in kb):
            models.append(model)

    result = all(query(model) for model in models)
    return result, models


# Propositional symbols
symbols = ['P', 'Q', 'R']

# Knowledge base:
# Q -> P, P -> not Q, Q OR R

kb = [
    lambda m: (not m['Q']) or m['P'],
    lambda m: (not m['P']) or (not m['Q']),
    lambda m: m['Q'] or m['R']
]

# Queries
queries = {
    "R": lambda m: m['R'],
    "R -> P": lambda m: (not m['R']) or m['P'],
    "Q -> R": lambda m: (not m['Q']) or m['R']
}

# Display truth table
print("P Q R | Q->P P->~Q Q|R | KB")
print("-" * 38)

for values in product([False, True], repeat=3):
    m = dict(zip(symbols, values))

    s1 = (not m['Q']) or m['P']
    s2 = (not m['P']) or (not m['Q'])
    s3 = m['Q'] or m['R']
    kb_true = s1 and s2 and s3

    print(
        int(m['P']), int(m['Q']), int(m['R']), "|",
        int(s1), int(s2), int(s3), "|", int(kb_true)
    )

# Check entailment
result, models = entails(kb, lambda m: True, symbols)

print("\nModels of KB:")
for model in models:
    print(model)

print("\nEntailment Results:")
for name, query in queries.items():
    result, _ = entails(kb, query, symbols)
    print(f"KB entails {name}: {result}")
```
