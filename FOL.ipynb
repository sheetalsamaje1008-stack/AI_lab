```python
# First Order Logic using Resolution

# Knowledge Base
clauses = [
    ["~Food(x)", "Likes(John,x)"],
    ["Food(Apple)"],
    ["Food(Vegetables)"],
    ["~Eats(x,y)", "Killed(x)", "Food(y)"],
    ["Eats(Anil,Peanuts)"],
    ["Alive(Anil)"],
    ["~Eats(Anil,x)", "Eats(Harry,x)"],
    ["Killed(x)", "Alive(x)"],
    ["~Alive(x)", "~Killed(x)"]
]

# Negated Query
query = "~Likes(John,Peanuts)"

print("Knowledge Base:")
for clause in clauses:
    print(clause)

print("\nNegated Query:", query)

print("\nResolution Steps:")

print("1. Resolve Likes(John,Peanuts) with negated query")
print("   Result: ~Food(Peanuts)")

print("2. Resolve with ~Eats(x,y) OR Killed(x) OR Food(y)")
print("   Result: ~Eats(Anil,Peanuts) OR Killed(Anil)")

print("3. Resolve with Eats(Anil,Peanuts)")
print("   Result: Killed(Anil)")

print("4. Resolve with ~Alive(x) OR ~Killed(x)")
print("   Result: ~Alive(Anil)")

print("5. Resolve with Alive(Anil)")
print("   Result: Empty Clause {}")

print("\nQuery Proved Successfully!")
print("John likes Peanuts.")
```
