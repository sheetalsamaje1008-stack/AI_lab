```python
import heapq

initial = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


def manhattan(state):
    distance = 0

    for i, tile in enumerate(state):

        if tile == 0:
            continue

        current_row = i // 3
        current_col = i % 3

        goal_index = goal.index(tile)
        goal_row = goal_index // 3
        goal_col = goal_index % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


def print_board(state):
    for i in range(0, 9, 3):
        print(" ".join(str(x) if x != 0 else "_"
                       for x in state[i:i+3]))
    print()


def neighbors(state):

    result = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:

        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_blank = nr * 3 + nc

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            result.append((tuple(new_state), move))

    return result


def a_star():

    # (f, g, state, path)
    priority_queue = []

    h = manhattan(initial)

    heapq.heappush(
        priority_queue,
        (h, 0, initial, [])
    )

    visited = set()

    while priority_queue:

        f, g, current, path = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        print("Move:", len(path))
        print("g =", g)
        print("h =", manhattan(current))
        print("f =", f)
        print_board(current)

        if current == goal:

            print("===== GOAL REACHED =====")
            print("Optimal number of moves:", g)
            print("Moves:", " -> ".join(path))

            return

        for next_state, move in neighbors(current):

            if next_state not in visited:

                new_g = g + 1
                new_h = manhattan(next_state)
                new_f = new_g + new_h

                heapq.heappush(
                    priority_queue,
                    (
                        new_f,
                        new_g,
                        next_state,
                        path + [move]
                    )
                )


a_star()
```
