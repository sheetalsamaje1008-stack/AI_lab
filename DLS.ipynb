```python
def print_puzzle(state):
    """Print the 8-puzzle state."""
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_next_states(state):
    """Generate all possible moves of the blank (0)."""
    next_states = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = {
        "UP": (-1, 0),
        "DOWN": (1, 0),
        "LEFT": (0, -1),
        "RIGHT": (0, 1)
    }

    for move, (dr, dc) in moves.items():
        new_row = row + dr
        new_col = col + dc

        # Check if the move is inside the puzzle
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col

            # Create a new state
            new_state = list(state)
            new_state[blank], new_state[new_blank] = (
                new_state[new_blank],
                new_state[blank]
            )

            next_states.append((tuple(new_state), move))

    return next_states


def depth_limited_search(state, goal, limit, path):
    """Depth-Limited Search for the 8-puzzle."""

    # 1. If state == goal
    if state == goal:
        return path

    # 2. If limit == 0
    if limit == 0:
        return None

    # 3 & 4. Generate and explore possible moves
    for next_state, move in get_next_states(state):

        # Avoid states already on the current path
        if next_state not in [s for s, m in path]:
            result = depth_limited_search(
                next_state,
                goal,
                limit - 1,
                path + [(next_state, move)]
            )

            if result is not None:
                return result

    # 5. Return FAILURE
    return None


# Initial state
initial_state = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

# Goal state
goal_state = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

# Depth limit
limit = 10

# Perform Depth-Limited Search
path = depth_limited_search(
    initial_state,
    goal_state,
    limit,
    [(initial_state, "START")]
)


# Display result
if path is not None:
    print("Solution found!")
    print("Number of moves:", len(path) - 1)
    print()

    for i, (state, move) in enumerate(path):
        print("Move:", move)
        print_puzzle(state)
else:
    print("No solution found within the depth limit.")
```
